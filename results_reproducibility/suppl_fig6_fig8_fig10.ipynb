# Supplementary Figures - Network-based analysis of genome-wide biobank data boosts discovery of genetic associations in psoriasis

### Abstract

Psoriasis is a common autoimmune disease with a strong genetic component. Targeting the _IL-23/IL-17_ pathways for the treatment of moderate-severe psoriasis has proven successful, which makes it a good benchmark for other drug discovery approaches. Genome Wide Association Studies (GWAS) identify genetic loci associated with a phenotype (e.g. disease). However, using identified loci to understand the disease mechanisms remains challenging. In response, gene networks methods that consider gene-gene relationships have been developed. In a previous study, we showed that the combination of network methods improved the power, reproducibility, and interpretability of those approaches on a small cohort of breast cancer patients. The goal of the present study is to benchmark the performance of the same combination of network methods on a psoriasis cohort from the UK biobank (with a particular interest on the _IL-23/IL-17_ pathways). By applying this method, we identified well known psoriasis genes, demonstrating the viability of our approach. Our method further identified a small set of genes of the MRPL family robustly associated with psoriasis, demonstrating that this approach can provide insights into the potential disease mechanisms. Our study also shows that our method is applicable to large-scale biobank data to produce novel findings, even in well-studied diseases with strong genetic signal.

## QQ plot

Observed vs expected −log10(P) of the association test, to check inflation of the test statistics.

[↑ Back to contents](#contents)

```python
import numpy as np
from scipy import stats

gwas = pd.read_csv("data/ukb22418_all_chr_b0_v2_filtered_qc-ed_no_mhc_ld.assoc", delim_whitespace=True)

p = gwas["P"].sort_values()

expected = -np.log10(np.arange(1, len(p)+1) / (len(p)+1))
observed = -np.log10(p)

plt.scatter(expected, observed, s=5)
plt.plot([0, max(expected)], [0, max(expected)])
plt.xlabel("Expected -log10(P)")
plt.ylabel("Observed -log10(P)")
plt.title("QQ plot")
#plt.savefig("qq_plot_nodeviation.pdf", dpi=150, bbox_inches="tight")
plt.show()
```

## UpSet plot — consensus networks by prior-knowledge network

Intersections between the consensus gene sets obtained with five PPI networks (bioGRID 128 genes, HINT 101, IID 258, IntAct 59, STRING 239). Cells below list genes found in ≥ 4 of the 5 sets and check the MRPL-family genes.

[↑ Back to contents](#contents)

```python
import pandas as pd
import matplotlib.pyplot as plt
from upsetplot import UpSet, from_contents
from matplotlib.text import Text
import numpy as np
```

```python
FILES = {
   # 'bioGRID' and 'baseline_bioGRID' are the same, but we keep both for labeling purposes
    "bioGRID": "data/consensus_44249.tsv",
    "HINT": "data/consensus_hint.tsv",
    "IID": "data/consensus_iid.tsv",
    "IntAct": "data/consensus_intact.tsv",
    "STRING": "data/consensus_string.tsv",
    # "finngen": "data/consensus_finngen.tsv",
    # "no_mhc_covars_bioGRID": "data/consensus_no_mhc_covars.tsv",
}

def load_set(path: str) -> set:
    """Read the first column of a TSV (skipping header) and return a set."""
    df = pd.read_csv(path, sep="\t", usecols=[0], header=0)
    return set(df.iloc[:, 0].dropna().astype(str))


sets = {name: load_set(path) for name, path in FILES.items()}

for name, s in sets.items():
    print(f"{name}: {len(s)} elements")


# ── 2.  UpSet plot (handles any number of sets) ───────────────────────────────
upset_data = from_contents(sets)

fig = plt.figure(figsize=(14, 6))
upset = UpSet(
    upset_data,
    subset_size="auto",
    show_counts=True,
    sort_by="cardinality",      # largest intersections first
    sort_categories_by=None,
)
# upset.plot(fig)
axes = upset.plot(fig)
axes["totals"].set_xlim(250, 0)
for ax in fig.axes:
  for text in ax.findobj(match=Text):
    x, y = text.get_position()
    if np.ndim(x) and np.size(x) == 1:
      x = np.asarray(x).item()
    if np.ndim(y) and np.size(y) == 1:
      y = np.asarray(y).item()
    text.set_position((x, y))
plt.suptitle("UpSet Plot - all intersections across different consensus based on biological networks", fontsize=14, y=1.02)
# plt.savefig("upset_plot.pdf", dpi=150, bbox_inches="tight")
plt.show()
# print("Saved: upset_plot.png")
```

```python
from itertools import combinations

# Elements present in at least 4 out of 5 sets
names = list(sets.keys())
at_least_4 = set()

for combo in combinations(names, 4):
    at_least_4 |= set.intersection(*(sets[n] for n in combo))

print(f"\nElements in at least 4 sets ({len(at_least_4)}):")
for element in sorted(at_least_4):
    print(f"  {element}")
```

```python
for element in sorted(at_least_4):
    present_in = [name for name, s in sets.items() if element in s]
    print(f"  {element}: found in {present_in}")
```

```python
elements_to_search = [
    "MRPL4",
    "MRPL24",
    "MRPL38",
    "DDX28",
    "CS",
    "VARS2",
    "MRPS18B",
]

print("\nSet membership for searched elements:")
for element in elements_to_search:
    present_in = [name for name, s in sets.items() if element in s]
    if present_in:
        print(f"  {element}: found in {present_in}")
    else:
        print(f"  {element}: not found in any set")
```

## Venn diagram — consensus networks by dataset

Overlap of the consensus gene sets from the baseline bioGRID analysis, FinnGen, and the analysis excluding the MHC region with covariates (24 genes shared by all three).

[↑ Back to contents](#contents)

```python
FILES = {
    # 'bioGRID' and 'baseline_bioGRID' are the same, but we keep both for labeling purposes
    "baseline_bioGRID": "data/consensus_44249.tsv",
    "finngen": "data/consensus_finngen.tsv",
    "no_mhc_covars_bioGRID": "data/consensus_no_mhc_covars.tsv",
}

sets = {name: load_set(path) for name, path in FILES.items()}

try:
    from venn import venn as venn3          # pip install venn
    fig, ax = plt.subplots(figsize=(10, 10))
    venn3(sets, ax=ax)
    ax.set_title("3-set Venn Diagram", fontsize=14)
    # plt.savefig("venn_3set.pdf", dpi=150, bbox_inches="tight")
    plt.show()
    # print("Saved: venn_3set.pdf")
except ImportError:
    print(
        "3-set Venn skipped – install the 'venn' package:  pip install venn\n"
        "The UpSet plot above already shows all intersections."
    )


# ── 4.  Print intersection sizes as a reference table ────────────────────────
from itertools import combinations

print("\n── Pairwise intersections ──")
names = list(sets.keys())
for r in range(2, len(names) + 1):
    for combo in combinations(names, r):
        common = set.intersection(*(sets[n] for n in combo))
        print(f"  {'∩'.join(combo)}: {len(common)}")
```
